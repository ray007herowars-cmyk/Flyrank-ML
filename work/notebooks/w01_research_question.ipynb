## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 content items. Three useful observations are:

- 65.7% of content items have a negative `trend_pct`.
- 54.4% have `trend_pct <= -20%`.
- Median impressions are 731 over 90 days compared with 139 in the last 30 days.

These are observed properties of the starter dataset, not evidence that ML will necessarily outperform a rule. They show that recent performance change is common enough to make content prioritization worth investigating over the next several weeks.

`trend_pct` is used here only as an observed outcome for understanding the dataset. It must not be used as a model feature, because the data documentation states that `is_declining_label` is derived from `trend_direction`, which itself is computed from `trend_pct`.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from pathlib import Path
import pandas as pd

# Support execution from either the repository root or this notebook's folder.
candidates = [
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("../../data/raw/content_refresh_anonymized.csv"),
]

data_path = next((p for p in candidates if p.exists()), None)

if data_path is None:
    raise FileNotFoundError(
        "Could not find data/raw/content_refresh_anonymized.csv"
    )

df = pd.read_csv(data_path)

negative_trend_pct = (df["trend_pct"] < 0).mean() * 100
severe_decline_pct = (df["trend_pct"] <= -20).mean() * 100
median_impressions_90d = df["impressions_90d"].median()
median_impressions_last_30d = df["impressions_last_30d"].median()

print(f"Rows: {len(df):,}")
print(f"Negative trend_pct: {negative_trend_pct:.1f}%")
print(f"trend_pct <= -20%: {severe_decline_pct:.1f}%")
print(f"Median impressions (90d): {median_impressions_90d:.0f}")
print(f"Median impressions (last 30d): {median_impressions_last_30d:.0f}")
```
