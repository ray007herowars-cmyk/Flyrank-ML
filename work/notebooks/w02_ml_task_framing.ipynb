## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Content performance can involve search demand, impressions, clicks, average position, content characteristics, intent, and other signals that may have different relationships across content items.

ML is worth testing if it can learn useful combinations of these signals and produce a better top-K review ranking than a simple rule.

However, I will not assume that ML wins. The baseline rule will be evaluated using the same metric, precision@K. If ML does not improve the ranking, a simpler rule may be the better solution.
